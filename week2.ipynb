```python
# Array filled with zeros

a = np.zeros((2,2))
print(a)

print("-----------------------")

# Array filled with ones

b = np.ones((3,3))
print(b)

print("-----------------------")

# Create an empty array with random content

c = np.empty((2,3))
print(c)

print("-----------------------")

# Create a constant array

d = np.full((2,2), 7)
print(d)

print("-----------------------")

## Create a 2x2 identity matrix

e = np.eye(2)
print(e)

print("-----------------------")

# Create an array with a range of elements

f = np.arange(4)
print(f)

print("-----------------------")

g = np.arange(1,10,3)
print(g)

print("-----------------------")

# Create an array with random values
h = np.random.random((2,2))
print(h)
```

### **Explore the differences between np.random.random(), np.random rand() and np.random.randn().**

| Function | Distribution | Range | Input Syntax (Shape) |
| :--- | :--- | :--- | :--- |
| **`np.random.random()`** | Uniform | $[0.0, 1.0)$ | Tuple/List: `(2, 3)` |
| **`np.random.rand()`** | Uniform | $[0.0, 1.0)$ | Separate Arguments: `2, 3` |
| **`np.random.randn()`** | Standard Normal (Gaussian) | $(-\infty, +\infty)$ | Separate Arguments: `2, 3` |



```python
x = np.random.random((2, 3))
print(x)
```

```python
y = np.random.rand(2, 3)
print(y)
```

```python
z = np.random.randn(2, 3)
print(z)
```

## NumPy Array Attributes

Key attributes to inspect the structure and memory layout of an array:

* **`.ndim`**: The number of axes (dimensions) in the array.
* **`.shape`**: A tuple indicating the size of the array in each dimension (e.g., rows, columns).
* **`.size`**: The total number of elements in the array (product of the shape).
* **`.dtype`**: The data type of the elements (e.g., `int64`, `float32`).

| Attribute | Description | Return Type | Example Output |
| :--- | :--- | :--- | :--- |
| **`.ndim`** | Number of dimensions | Integer | `2` |
| **`.shape`** | Dimensions (rows, cols) | Tuple | `(3, 4)` |
| **`.size`** | Total element count | Integer | `12` |
| **`.dtype`** | Data type of elements | Object | `float64` |

```python
# Creating a 3-Dimensional array (2 layers, 3 rows, 4 columns)

arr = np.random.randint(10, size=(2, 3, 4))
print(arr)
```
