```python
input_string = "The capital of Sri Lanka is Colombo"

# Initialize counters
lowercase_count = 0
uppercase_count = 0

# Iterate through the string character by character
for char in input_string:
    # Check if the character is an uppercase letter
    if char.isupper():
        uppercase_count += 1
    # Check if the character is a lowercase letter
    elif char.islower():
        lowercase_count += 1
    # Ignore spaces, punctuation, and other characters (optional, but good practice)
    else:
        continue

# Print the results
print(f"Original String: '{input_string}'")
print("-" * 30)
print(f"Total Uppercase Letters: {uppercase_count}")
print(f"Total Lowercase Letters: {lowercase_count}")
```

```python
input_string = "The capital of Sri Lanka is Colombo"

# Initialize counters
lowercase_count = 0
uppercase_count = 0

for char in input_string:
    if char.isupper():
        uppercase_count += 1
    elif char.islower():
        lowercase_count += 1
    
    
print(f"Original String: '{input_string}'")
print(lowercase_count)
print(uppercase_count)
print(lowercase_count + uppercase_count)
```

```python
def count_case(input_string):
    lowercase_count = 0
    uppercase_count = 0

    for char in input_string:
        if char.isupper():
            uppercase_count += 1
        elif char.islower():
            lowercase_count += 1

    return lowercase_count, uppercase_count
```

```python
name = 'Sukitharathnayake'
lowercase, uppercase = count_case(name)
print(f"Lowercase letters: {lowercase}")
print(f"Uppercase letters: {uppercase}")
```

```python
x_cors=[1,3,1.5,2,2.5,10,11,12,9,14]
y_cors=[2,2.5,1,0.5,3,9,11,12,10,10.5]

plt.scatter(x_cors, y_cors)
plt.xlabel('Time spent at a parking lot')
plt.ylabel('The parking cost')
plt.title('Scatter Plot of Given Coordinates')
plt.show()
```

```python
x_cors=[1,3,1.5,2,2.5,10,11,12,9,14]
y_cors=[2,2.5,1,0.5,3,9,11,12,10,10.5]

c1_X, c1_Y = x_cors[:5], y_cors[:5]
c2_X, c2_Y = x_cors[5:], y_cors[5:]

plt.scatter(c1_X, c1_Y, color='blue', label='Cluster 1')
plt.scatter(c2_X, c2_Y, color='red', label='Cluster 2')
plt.xlabel('Time spent at a parking lot')
plt.ylabel('The parking cost')
plt.title('Scatter Plot with Clusters')
plt.legend()
plt.show()
```

```python
x = np.column_stack((x_cors, y_cors))
print(x)
```

```python
dist_matrix = squareform(pdist(x, 'euclidean'))
```

```python
print(dist_matrix)
```

```python
plt.imshow(dist_matrix, cmap='viridis', origin='upper', interpolation='nearest')
plt.colorbar(label='Euclidean Distance')
plt.title('Pairwise Distance Heatmap')
plt.xlabel('Point Index')
plt.ylabel('Point Index')
```
